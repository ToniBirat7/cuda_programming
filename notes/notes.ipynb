`CUDA` is a parallel computing platform and programming model developed by NVIDIA that allows developers to utilize the power of NVIDIA GPUs for general-purpose computing tasks. It provides a way to write programs that can run on the GPU, which can significantly accelerate certain types of computations compared to running them on a CPU.

It's an extentions of C/C++ programming language, and it provides a set of APIs and libraries that enable developers to write code that can be executed on the GPU. 

## **SIMT Model**

In classical execution, say when adding two vectors:

In a **classical (sequential) execution model**, a single processing unit computes one element at a time, in order.

Let:

$$
\mathbf{a} = [1, 2, 3, 4, 5], \quad
\mathbf{b} = [10, 20, 30, 40, 50]
$$

We want:

$$
\mathbf{c} = \mathbf{a} + \mathbf{b}
$$

with:

$$
c_i = a_i + b_i,\quad i = 0,1,2,3,4
$$

So the CPU performs additions **sequentially**:

$$
\begin{aligned}
c_0 &= a_0 + b_0 = 1 + 10 = 11 \\
c_1 &= a_1 + b_1 = 2 + 20 = 22 \\
c_2 &= a_2 + b_2 = 3 + 30 = 33 \\
c_3 &= a_3 + b_3 = 4 + 40 = 44 \\
c_4 &= a_4 + b_4 = 5 + 50 = 55
\end{aligned}
$$

Final result:

$$
\mathbf{c} = [11, 22, 33, 44, 55]
$$

This is the **sequential model**: one instruction stream processes one element after another.

Now in the **SIMT (Single Instruction, Multiple Threads)** model used by CUDA, multiple threads execute the same instruction on different data elements in parallel.

In the SIMT model, we launch multiple threads to perform additions concurrently. Each thread executes the same instruction (vector addition), but on a different index.

### **Threads**

In CUDA, we launch a kernel (a function that runs on the GPU) with multiple threads. Each thread computes one element of the result vector `c`.

- Threads are the lowest granularity of execution in CUDA.

- Execute the same instruction (vector addition) on different data elements in parallel.

For vectors `a` and `b` of size 5, we can launch 5 threads:

$$
\begin{aligned}
\text{Thread 0: } & c_0 = a_0 + b_0 = 1 + 10 = 11 \\
\text{Thread 1: } & c_1 = a_1 + b_1 = 2 + 20 = 22 \\
\text{Thread 2: } & c_2 = a_2 + b_2 = 3 + 30 = 33 \\
\text{Thread 3: } & c_3 = a_3 + b_3 = 4 + 40 = 44 \\
\text{Thread 4: } & c_4 = a_4 + b_4 = 5 + 50 = 55
\end{aligned}
$$

So the final output is still:

$$
\mathbf{c} = [11, 22, 33, 44, 55]
$$

but the computation is done in parallel, which is the key idea behind CUDA's SIMT execution model.

**Warps(SIMT)**

In CUDA, threads are organized into groups called **warps**. A warp is a group of 32 threads that execute instructions in lockstep. This means that all threads in a warp execute the same instruction at the same time, but on different data.

- Lowest schedulable entiry in CUDA is a warp (32 threads).

- Executes instructions in lockstep: all threads in a warp execute the same instruction simultaneously. 

  - Not every thread needs to execute the same instruction, but if they do, they execute it together.

For example, if we have 64 threads, they will be organized into 2 warps:

- Warp 0: Threads 0-31  

- Warp 1: Threads 32-63

## **Thread Blocks and Grids**

In CUDA, threads are further organized into **thread blocks** and **grids** to manage larger computations.

### **Thread Blocks**

A thread block is a group of threads that can cooperate with each other by sharing data through shared memory and synchronizing their execution. Each block can contain up to 1024 threads (depending on the GPU architecture).

If we say i want 256 threads, we can organize them into 8 blocks of 32 threads each:

- Block 0: Threads 0-31 (Warp 0)

- Block 1: Threads 32-63 (Warp 1)

- Block 2: Threads 64-95 (Warp 2)

- Block 3: Threads 96-127 (Warp 3)

- Block 4: Threads 128-159 (Warp 4)

So now these 8 blocks of 32 threads is called a `thread block`. Each block can execute independently, and threads within the same block can communicate and synchronize with each other.

> All of the `wraps` gets collectively called a `thread block`.

> It is the lowest programmable unit in CUDA.

> Block is assigned to a sinle `Shader Core` (or SM) for execution in GPU

> Thead blocks are 3 dimensional, so we can have blocks of size (8, 8, 8) which means 512 threads per block.

### **Grids**

A grid is a collection of thread blocks. When we launch a kernel, we specify the number of blocks and the number of threads per block. The grid can be one-dimensional, two-dimensional, or three-dimensional.

> `Grid` represents how a problem is mapped to the GPU

> Part of the GPU launch configuration, where we specify the number of blocks and threads per block.
