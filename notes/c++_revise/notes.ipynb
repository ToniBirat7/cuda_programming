## **Pointer Notes**

> Special Notes

- Default size of a pointer variable is 4 bytes in 32-bit systems and 8 bytes in 64-bit systems, regardless of the type of data it points to.

- When `*` asterisk is used in a `Declaration` it is used to declare a pointer variable. For example, `int* ptr;` declares a pointer to an integer.

- When `*` asterisk is used in an `Expression` it is used to dereference a pointer, which means accessing the value stored at the memory address that the pointer points to. For example, `int value = *ptr;` retrieves the value that `ptr` points to.

<hr>

- We use `new` and `malloc` to allocate memory dynamically in `Heap`. The `new` operator is used in C++ and returns a pointer to the allocated memory. Also, the `new` operator automatically returns the correct type of pointer based on the type of variable being allocated. 

For example:

```cpp
int* ptr = new int; // allocates memory for a single integer using new, automatically returns an int pointer, we don't have to use void pointer and cast it to int pointer
*ptr = 10; // assigns a value to the allocated memory
```

- When we create a variable using `new` we need to manually delete the allocated memory using `delete` to avoid memory leaks. 

- while `malloc` is used in C and also returns a pointer to the allocated memory. For example:

```cpp

// Malloc example in C
int* ptr = (int*)malloc(sizeof(int)); // allocates memory for a single integer, returns a void pointer, so we need to cast it to an int pointer
*ptr = 10; // assigns a value to the allocated memory
```

<hr>

Pointers are variables that store memory addresses. They allow us to directly access and manipulate memory, which can lead to more efficient code. In C/C++, we can declare a pointer using the `*` operator. For example:

```cpp
int* ptr; // declares a pointer to an integer
```
We can also initialize a pointer to point to a specific variable:

```cpp
int x = 10;
int* ptr = &x; // ptr now points to the memory address of x
```

To access the value stored at the memory address that a pointer points to, we can use the dereference operator `*`. For example:

```cpp
int value = *ptr; // value now holds the value of x, which is 10
```



## **Double Pointers**

Double pointers are pointers that point to other pointers. They are declared using two `*` operators. For example:

```cpp
#include <stdio.h>
using namespace std;

int main() {
    int x = 42;
    int* ptr = &x; // Stores the pointer to x 
    int** ptr2 = &ptr; // 
    int***ptr3 = &ptr2;

    printf("Value: %d\n", ***ptr3);
}
```

To understand how this code works and why we need multiple asterisks (`*`, `**`, `***`), it helps to visualize how computer memory works. 

In C and C++, every variable has a **value** and a **memory address** (where that value is stored). A pointer is simply a special variable that stores the memory address of another variable.

---

### Step 1: Visualizing the Memory

Let's look at what happens in memory step-by-step when your code runs. We will use hypothetical memory addresses (like `0x100`, `0x200`, etc.) to make it clear:

| Variable | Type | Value (What it stores) | Address of this variable | What it points to |
| :--- | :--- | :--- | :--- | :--- |
| **`x`** | `int` | `42` | `0x100` | *(An actual integer value)* |
| **`ptr`** | `int*` | `0x100` *(Address of x)* | `0x200` | Points to `x` |
| **`ptr2`** | `int**` | `0x200` *(Address of ptr)* | `0x300` | Points to `ptr` |
| **`ptr3`** | `int***` | `0x300` *(Address of ptr2)* | `0x400` | Points to `ptr2` |

---

### Step 2: Why can't we use a single asterisk (`*`) for everything?

In C and C++, the compiler is strictly typed. It needs to know exactly what kind of data is at the destination of a pointer. 

* **`int*`** means: *"This variable stores the address of an `int`."*
* **`int**`** means: *"This variable stores the address of an `int*` (a pointer to an integer)."*
* **`int***`** means: *"This variable stores the address of an `int**` (a pointer to a pointer to an integer)."*

If you tried to write:
```cpp
int* ptr2 = &ptr; // Error!
```
The compiler would reject this. Why? 
Because `ptr` is an `int*`. The address of `ptr` (`&ptr`) is of type `int**`. If you try to store it in a single-pointer `int* ptr2`, the compiler gets confused. It thinks `ptr2` is pointing directly to an integer, when in reality it is pointing to another pointer. 

Each extra asterisk tells the compiler, **"You need to take one more jump through memory to reach the actual integer."**

---

### Step 3: How Dereferencing Works (`***ptr3`)

The asterisk (`*`) has two different meanings depending on where it is used:
1. **In a declaration** (like `int* ptr`), it means "this is a pointer type."
2. **In an expression** (like `*ptr`), it is the dereference operator, which means "go to the address stored in this pointer and grab the value."

When you print `***ptr3`, you are dereferencing it three times. Think of it like peeling layers of an onion:

1. **`ptr3`** holds the address `0x300` (which is where `ptr2` lives).
2. **`*ptr3`** (First dereference): Go to `0x300`. The value stored there is `0x200` (which is the address of `ptr`).
3. **`**ptr3`** (Second dereference): Go to `0x200`. The value stored there is `0x100` (which is the address of `x`).
4. **`***ptr3`** (Third dereference): Go to `0x100`. The value stored there is `42`.

This is why the output of your program is `42`. If you only used `*ptr3`, it would print the memory address stored in `ptr2`. If you used `**ptr3`, it would print the memory address stored in `ptr`. To get the final integer, you must dereference it three times.



## **Pointer Arithmetic**

Pointer arithmetic allows us to perform operations on pointers, such as incrementing or decrementing them. When we increment a pointer, it moves to the next memory address based on the size of the data type it points to. For example:

```cpp
ptr++; // moves the pointer to the next integer memory address
```

### **Pointers and Arrays**

In C/C++, arrays and pointers are closely related. An array name can be treated as a pointer to the first element of the array. For example:

```cpp
int arr[5] = {1, 2, 3, 4, 5};
int* ptr = arr; // ptr points to the first element of the array
```

If we increase the pointer `ptr` by 1 i.e. `ptr++`, it will point to the next element of the array, which is `2`. This is because the pointer arithmetic takes into account the size of the data type. In this case, since `ptr` is an `int*`, incrementing it will move it next `4` bytes (the size of an integer) to point to the next element in the array.

As arrays are layout in contiguous memory locations, we can use pointer arithmetic to traverse through the array elements. For example:

```cpp
for (int i = 0; i < 5; i++) {
    printf("%d ", *ptr); // prints the current element
    ptr++; // moves to the next element
}
```

### **Pointers and Functions**

Pointers can also be used to pass arguments to functions by reference. This allows the function to modify the original variable. For example:

```cpp
void increment(int* num) {
    (*num)++; // increments the value at the memory address pointed to by num
}

int main() {
    int x = 5;
    increment(&x); // passes the address of x to the function
    // x is now 6
}
```

### **Pointers and Dynamic Memory Allocation**

> Normally when we compile `C++` program, all the memory is allocated at compile time, and the size of the memory is fixed. However, sometimes we may need to allocate memory dynamically at runtime, which is where pointers come in handy. This is called `Dynamic Memory Allocation`.

Pointers are essential for dynamic memory allocation, which allows us to allocate memory at runtime. In C/C++, we can use the `new` operator to allocate memory for a variable or an array. For example:

```cpp
int* ptr = new int; // allocates memory for a single integer
*ptr = 10; // assigns a value to the allocated memory
int* arr = new int[5]; // allocates memory for an array of 5 integers
```

> If we create a pointer using `new`, it is stored in the `Heap` memory, and even if it is declared inside a function, it will not be destroyed when the function ends. We need to manually free the allocated memory using the `delete` operator to avoid memory leaks:

```cpp
delete ptr; // frees the memory allocated for the single integer
delete[] arr; // frees the memory allocated for the array of integers
```

## **Void Pointers**

A void pointer, also known as a generic pointer, is a special type of pointer that can point to any data type. It is declared using the `void*` syntax. For example:

```bash
void* ptr; // declares a void pointer
```

> Previously when we created a pointer, we had to specify the data type it pointed to. However, with void pointers, we can point to any data type without specifying it. This makes void pointers very flexible and useful in situations where we want to write functions that can work with different types of data.

To use a void pointer, we need to cast it to the appropriate data type before dereferencing it. For example:

```cpp
int x = 10;
void* ptr = &x; // ptr points to the memory address of x  

int value = *(int*)ptr; // casts ptr to an int pointer and dereferences it to get the value of x
```

Note: 

`malloc` returns a void pointer, so we need to cast it to the appropriate data type when using it in C++. In C, the cast is not necessary. For example:

```cpp
int* ptr = (int*)malloc(sizeof(int)); // allocates memory for a single integer
*ptr = 10; // assigns a value to the allocated memory
```

> Note: `malloc` returns a void pointer, so we need to cast it to the appropriate data type when using it in C++. In C, the cast is not necessary. For example:

```cpp
int* ptr = (int*)malloc(sizeof(int)); // allocates memory for a single integer
*ptr = 10; // assigns a value to the allocated memory
```

## **Null Pointers**

Null pointers are pointers that do not point to any valid memory address. They are often used to indicate that a pointer is not currently pointing to anything. In C/C++, we can assign a null pointer using the `nullptr` keyword (introduced in C++11) or the `NULL` macro (in C). For example:

```cpp
int* ptr = nullptr; // ptr is a null pointer
```
We cannot dereference a null pointer, as it does not point to a valid memory address. Attempting to do so will result in undefined behavior, which can lead to crashes or other unexpected results. Therefore, it is important to check if a pointer is null before dereferencing it:

```cpp
if (ptr != nullptr) {
    // safe to dereference ptr
} else {
    // ptr is null, handle accordingly
}
```

To derefence a null pointer, we can use the `*` operator, but it will lead to undefined behavior. For example:

```cpp
int value = *ptr; // this will cause undefined behavior since ptr is a null pointer
```

Therefore, to successfully dereference a null pointer, we need to type cast it to `void*` first, and then cast it back to the appropriate data type. For example:

```cpp
int value = *(int*)(void*)ptr; // this will still cause undefined behavior, but it demonstrates how to cast a null pointer to void* and back to int*
```

## **Matrix with Arrays and Pointers**

We can create a matrix (2D array) using arrays and pointers in C/C++. A matrix is essentially an array of arrays. We can declare a 2D array using the following syntax:

```cpp

void matrix_2d_array() {

    int arr[][4] = {{1,2,3,4}, {5,6,7,8}};

    for (int i = 0; i < 2; i++) {
        for (int j = 0; j <=3; j++) {
            printf("%d ", arr[i][j]);
        }
        printf("\n");
    }
}
```

Also, we can create a matrix using pointers. This involves creating an array of pointers, where each pointer points to a row of the matrix. For example:

```cpp
void matrix() {
    int arr1[] = {1,2,3,4};
    int arr2[] = {5,6,7,8};

    int* ptr1 = arr1; // Pointer to first array
    int* ptr2 = arr2; // Pointer to second array

    int* matrix[] = {ptr1, ptr2}; // Array of pointers that represents a matrix

    for (int i = 0; i < 2; i++) {
        for (int j = 0; j <=3; j++) {
            printf("%d ", *matrix[i]++);
        }
        printf("\n");
    }
}
```

We've just successfully coded the exact difference between a **Contiguous 2D Array** and an **Array of Pointers**. 

Even though both functions print the exact same output:
```text
1 2 3 4 
5 6 7 8 
```
Under the hood, their memory layouts and mechanics are completely different. **One of these is great for CUDA, and the other is a nightmare.**

Here is the exact difference between the two.

---

### 1. `matrix_2d_array()`: The Contiguous Array (The Good Way)

```cpp
int arr[][4] = {{1,2,3,4}, {5,6,7,8}};
```

**How memory looks:**
Because you declared it as a true 2D array, the compiler guarantees that all 8 numbers are placed side-by-side in one solid, unbroken block of memory (**Contiguous Memory**).

```text
Memory Address:  [0x00] [0x04] [0x08] [0x0C] [0x10] [0x14] [0x18] [0x1C]
Data:            [ 1  ] [ 2  ] [ 3  ] [ 4  ] [ 5  ] [ 6  ] [ 7  ] [ 8  ]
```

**Why it matters for CUDA:**

Because it is one solid block, you can copy this entire matrix to your RTX 3060 with exactly **one** `cudaMemcpy` call. The hardware picks up the block starting at `0x00` and ships all 32 bytes (8 ints * 4 bytes) over the PCIe bus instantly.

---

### 2. `matrix()`: The Array of Pointers (The Bad Way for CUDA)

```cpp
int arr1[] = {1,2,3,4};
int arr2[] = {5,6,7,8};
int* matrix[] = {ptr1, ptr2};
```

**How memory looks:**
You did not create a 2D array here. You created **three separate 1D arrays**. 

The memory is fragmented. `arr1` and `arr2` could be located anywhere in your CPU RAM, and `matrix` is just an array holding their addresses.

```text
[0x50] matrix: [ ptr to 0x10 ] [ ptr to 0x80 ]
                     |               |
                     v               v
[0x10] arr1:   [ 1 ][ 2 ][ 3 ][ 4 ]  |
                                     |
[0x80] arr2:                   [ 5 ][ 6 ][ 7 ][ 8 ]
```

**Why it is a nightmare for CUDA:**

If you try to copy `matrix` to the GPU using `cudaMemcpy`, the GPU will only receive the **pointers** (`0x10` and `0x80`). But those are CPU memory addresses! 

If the GPU tries to read `0x10` to find the number `1`, the program will crash because the GPU does not have access to CPU RAM.

To send this structure to the GPU, you would have to:

1. `cudaMalloc` space for `arr1` and `cudaMemcpy` it.
2. `cudaMalloc` space for `arr2` and `cudaMemcpy` it.
3. Create a new array of pointers *on the GPU*, pointing to the new GPU addresses.

*This process is called a "Deep Copy" and it is incredibly slow and complex.*

---

### 3. A Sneaky Bug in your `matrix()` loop
Take a very close look at this line of your code:
```cpp
printf("%d ", *matrix[i]++);
```
This is a very clever C++ pointer trick, but it is **destructive**. 

Because of C++ operator precedence, `matrix[i]++` happens first. This permanently modifies the pointer stored inside your `matrix` array, pushing it forward by 1 memory slot. Then, the `*` dereferences it to print the number.

It works perfectly the first time you run it. But if you try to print the matrix a **second time** inside the same function:

```cpp
matrix(); // If you looped through it again inside this function...
```

It would print garbage values or crash your program! Why? Because your pointers `ptr1` and `ptr2` were moved to the end of the arrays during the first loop, and now point to empty space in your computer's RAM.

In `matrix_2d_array()`, you used `arr[i][j]`. This simply calculates an offset (the Golden Formula: `row * 4 + col`) and reads the value, leaving the original array completely untouched and safe to read again.

### Summary

* When a 2D array is created with `arr[][4]`, the compiler allocates a single block of contiguous memory for all the elements. Basically flat. This is perfect for hardware, caches, and GPUs. And to access the element the compiler uses the Golden Formula: `(row * COLS) + col` for the indexing.

* `matrix_2d_array()` = **Contiguous memory**. Perfect for hardware, caches, and GPUs. (Though in CUDA, we do it flat: `std::vector<int> arr = {1,2,3,4,5...}`).

* `matrix()` = **Array of pointers**. Fragmented memory. Good if you want a "jagged array" (where row 1 has 4 items, but row 2 has 99 items), but terrible for high-performance GPU computing.

## **Custom Types**

In C++, we use `size_t` to represent sizes and counts. 

- It is an `unsigned integer of long` type that is guaranteed to be large enough to hold the size of any object in memory.

- It has size of `64` bits on a 64-bit system, and 32 bits on a 32-bit system.

**Understand with example:**

If we've to create an array of 10 integers, we can use `size_t` to represent the size of the array:

```cpp
size_t size = 10; // size of the array
int* arr = new int[size]; // allocate memory for an array of 10 integers
```

**Struct**

- Default size of `Struct` is the sum of the sizes of its members, plus any padding added for alignment. The size can vary based on the types of members and the compiler's alignment requirements.

Another way of creating custom types in C++ is using `Structs`. A struct is a user-defined data type that can hold multiple variables of different types. 

For example:

```cpp
struct Point {
    int x; // x-coordinate
    int y; // y-coordinate
};

Point p1; // create an instance of Point
p1.x = 10; // assign value to x
p1.y = 20; // assign value to y
```
In this example, we defined a struct called `Point` that has two integer members, `x` and `y`. We then created an instance of `Point` called `p1` and assigned values to its members.

> Difference between `Structs` and `Classes` in C++:

- In C++, the only difference between a `struct` and a `class` is the default access level. In a `struct`, members are public by default, while in a `class`, members are private by default.
