- When `*` asterisk is used in an `Expression` it is used to dereference a pointer, which means accessing the value stored at the memory address that the pointer points to. For example, `int value = *ptr;` retrieves the value that `ptr` points to.

Pointers are variables that store memory addresses. They allow us to directly access and manipulate memory, which can lead to more efficient code. In C/C++, we can declare a pointer using the `*` operator. For example:

```cpp
int* ptr; // declares a pointer to an integer
```
We can also initialize a pointer to point to a specific variable:

```cpp
int x = 10;
int* ptr = &x; // ptr now points to the memory address of x
```



## **Double Pointers**

Double pointers are pointers that point to other pointers. They are declared using two `*` operators. For example:

```cpp
#include <stdio.h>
using namespace std;

int main() {
    int x = 42;
    int* ptr = &x; // Stores the pointer to x 
    int** ptr2 = &ptr; // 
    int***ptr3 = &ptr2;

    printf("Value: %d\n", ***ptr3);
}
```

To understand how this code works and why we need multiple asterisks (`*`, `**`, `***`), it helps to visualize how computer memory works. 

In C and C++, every variable has a **value** and a **memory address** (where that value is stored). A pointer is simply a special variable that stores the memory address of another variable.

---

### Step 1: Visualizing the Memory

Let's look at what happens in memory step-by-step when your code runs. We will use hypothetical memory addresses (like `0x100`, `0x200`, etc.) to make it clear:

| Variable | Type | Value (What it stores) | Address of this variable | What it points to |
| :--- | :--- | :--- | :--- | :--- |
| **`x`** | `int` | `42` | `0x100` | *(An actual integer value)* |
| **`ptr`** | `int*` | `0x100` *(Address of x)* | `0x200` | Points to `x` |
| **`ptr2`** | `int**` | `0x200` *(Address of ptr)* | `0x300` | Points to `ptr` |
| **`ptr3`** | `int***` | `0x300` *(Address of ptr2)* | `0x400` | Points to `ptr2` |

---

### Step 2: Why can't we use a single asterisk (`*`) for everything?

In C and C++, the compiler is strictly typed. It needs to know exactly what kind of data is at the destination of a pointer. 

* **`int*`** means: *"This variable stores the address of an `int`."*
* **`int**`** means: *"This variable stores the address of an `int*` (a pointer to an integer)."*
* **`int***`** means: *"This variable stores the address of an `int**` (a pointer to a pointer to an integer)."*

If you tried to write:
```cpp
int* ptr2 = &ptr; // Error!
```
The compiler would reject this. Why? 
Because `ptr` is an `int*`. The address of `ptr` (`&ptr`) is of type `int**`. If you try to store it in a single-pointer `int* ptr2`, the compiler gets confused. It thinks `ptr2` is pointing directly to an integer, when in reality it is pointing to another pointer. 

Each extra asterisk tells the compiler, **"You need to take one more jump through memory to reach the actual integer."**

---

### Step 3: How Dereferencing Works (`***ptr3`)

The asterisk (`*`) has two different meanings depending on where it is used:
1. **In a declaration** (like `int* ptr`), it means "this is a pointer type."
2. **In an expression** (like `*ptr`), it is the dereference operator, which means "go to the address stored in this pointer and grab the value."

When you print `***ptr3`, you are dereferencing it three times. Think of it like peeling layers of an onion:

1. **`ptr3`** holds the address `0x300` (which is where `ptr2` lives).
2. **`*ptr3`** (First dereference): Go to `0x300`. The value stored there is `0x200` (which is the address of `ptr`).
3. **`**ptr3`** (Second dereference): Go to `0x200`. The value stored there is `0x100` (which is the address of `x`).
4. **`***ptr3`** (Third dereference): Go to `0x100`. The value stored there is `42`.

This is why the output of your program is `42`. If you only used `*ptr3`, it would print the memory address stored in `ptr2`. If you used `**ptr3`, it would print the memory address stored in `ptr`. To get the final integer, you must dereference it three times.



## **Pointer Arithmetic**

Pointer arithmetic allows us to perform operations on pointers, such as incrementing or decrementing them. When we increment a pointer, it moves to the next memory address based on the size of the data type it points to. For example:

```cpp
ptr++; // moves the pointer to the next integer memory address
```

### **Pointers and Arrays**

In C/C++, arrays and pointers are closely related. An array name can be treated as a pointer to the first element of the array. For example:

```cpp
int arr[5] = {1, 2, 3, 4, 5};
int* ptr = arr; // ptr points to the first element of the array
```

If we increase the pointer `ptr` by 1 i.e. `ptr++`, it will point to the next element of the array, which is `2`. This is because the pointer arithmetic takes into account the size of the data type. In this case, since `ptr` is an `int*`, incrementing it will move it next `4` bytes (the size of an integer) to point to the next element in the array.

As arrays are layout in contiguous memory locations, we can use pointer arithmetic to traverse through the array elements. For example:

```cpp
for (int i = 0; i < 5; i++) {
    printf("%d ", *ptr); // prints the current element
    ptr++; // moves to the next element
}
```

### **Pointers and Functions**

Pointers can also be used to pass arguments to functions by reference. This allows the function to modify the original variable. For example:

```cpp
void increment(int* num) {
    (*num)++; // increments the value at the memory address pointed to by num
}

int main() {
    int x = 5;
    increment(&x); // passes the address of x to the function
    // x is now 6
}
```

### **Pointers and Dynamic Memory Allocation**

> Normally when we compile `C++` program, all the memory is allocated at compile time, and the size of the memory is fixed. However, sometimes we may need to allocate memory dynamically at runtime, which is where pointers come in handy. This is called `Dynamic Memory Allocation`.

Pointers are essential for dynamic memory allocation, which allows us to allocate memory at runtime. In C/C++, we can use the `new` operator to allocate memory for a variable or an array. For example:

```cpp
int* ptr = new int; // allocates memory for a single integer
*ptr = 10; // assigns a value to the allocated memory
int* arr = new int[5]; // allocates memory for an array of 5 integers
```

> If we create a pointer using `new`, it is stored in the `Heap` memory, and even if it is declared inside a function, it will not be destroyed when the function ends. We need to manually free the allocated memory using the `delete` operator to avoid memory leaks:

```cpp
delete ptr; // frees the memory allocated for the single integer
delete[] arr; // frees the memory allocated for the array of integers
```

## **Null Pointers**

Null pointers are pointers that do not point to any valid memory address. They are often used to indicate that a pointer is not currently pointing to anything. In C/C++, we can assign a null pointer using the `nullptr` keyword (introduced in C++11) or the `NULL` macro (in C). For example:

```cpp
int* ptr = nullptr; // ptr is a null pointer
```
We cannot dereference a null pointer, as it does not point to a valid memory address. Attempting to do so will result in undefined behavior, which can lead to crashes or other unexpected results. Therefore, it is important to check if a pointer is null before dereferencing it:

```cpp
if (ptr != nullptr) {
    // safe to dereference ptr
} else {
    // ptr is null, handle accordingly
}
```

To derefence a null pointer, we can use the `*` operator, but it will lead to undefined behavior. For example:

```cpp
int value = *ptr; // this will cause undefined behavior since ptr is a null pointer
```

Therefore, to successfully dereference a null pointer, we need to type cast it to `void*` first, and then cast it back to the appropriate data type. For example:

```cpp
int value = *(int*)(void*)ptr; // this will still cause undefined behavior, but it demonstrates how to cast a null pointer to void* and back to int*
```

## **Matrix with Arrays and Pointers**

We can create a matrix (2D array) using arrays and pointers in C/C++. A matrix is essentially an array of arrays. We can declare a 2D array using the following syntax:

```cpp

void matrix_2d_array() {

    int arr[][4] = {{1,2,3,4}, {5,6,7,8}};

    for (int i = 0; i < 2; i++) {
        for (int j = 0; j <=3; j++) {
            printf("%d ", arr[i][j]);
        }
        printf("\n");
    }
}
```

Also, we can create a matrix using pointers. This involves creating an array of pointers, where each pointer points to a row of the matrix. For example:

```cpp
void matrix() {
    int arr1[] = {1,2,3,4};
    int arr2[] = {5,6,7,8};

    int* ptr1 = arr1; // Pointer to first array
    int* ptr2 = arr2; // Pointer to second array

    int* matrix[] = {ptr1, ptr2}; // Array of pointers that represents a matrix

    for (int i = 0; i < 2; i++) {
        for (int j = 0; j <=3; j++) {
            printf("%d ", *matrix[i]++);
        }
        printf("\n");
    }
}
```

We've just successfully coded the exact difference between a **Contiguous 2D Array** and an **Array of Pointers**. 

Even though both functions print the exact same output:
```text
1 2 3 4 
5 6 7 8 
```
Under the hood, their memory layouts and mechanics are completely different. **One of these is great for CUDA, and the other is a nightmare.**

Here is the exact difference between the two.

---

### 1. `matrix_2d_array()`: The Contiguous Array (The Good Way)

```cpp
int arr[][4] = {{1,2,3,4}, {5,6,7,8}};
```

**How memory looks:**
Because you declared it as a true 2D array, the compiler guarantees that all 8 numbers are placed side-by-side in one solid, unbroken block of memory (**Contiguous Memory**).

```text
Memory Address:  [0x00] [0x04] [0x08] [0x0C] [0x10] [0x14] [0x18] [0x1C]
Data:            [ 1  ] [ 2  ] [ 3  ] [ 4  ] [ 5  ] [ 6  ] [ 7  ] [ 8  ]
```

**Why it matters for CUDA:**

Because it is one solid block, you can copy this entire matrix to your RTX 3060 with exactly **one** `cudaMemcpy` call. The hardware picks up the block starting at `0x00` and ships all 32 bytes (8 ints * 4 bytes) over the PCIe bus instantly.

---

### 2. `matrix()`: The Array of Pointers (The Bad Way for CUDA)

```cpp
int arr1[] = {1,2,3,4};
int arr2[] = {5,6,7,8};
int* matrix[] = {ptr1, ptr2};
```

**How memory looks:**
You did not create a 2D array here. You created **three separate 1D arrays**. 

The memory is fragmented. `arr1` and `arr2` could be located anywhere in your CPU RAM, and `matrix` is just an array holding their addresses.

```text
[0x50] matrix: [ ptr to 0x10 ] [ ptr to 0x80 ]
                     |               |
                     v               v
[0x10] arr1:   [ 1 ][ 2 ][ 3 ][ 4 ]  |
                                     |
[0x80] arr2:                   [ 5 ][ 6 ][ 7 ][ 8 ]
```

**Why it is a nightmare for CUDA:**

If you try to copy `matrix` to the GPU using `cudaMemcpy`, the GPU will only receive the **pointers** (`0x10` and `0x80`). But those are CPU memory addresses! 

If the GPU tries to read `0x10` to find the number `1`, the program will crash because the GPU does not have access to CPU RAM.

To send this structure to the GPU, you would have to:

1. `cudaMalloc` space for `arr1` and `cudaMemcpy` it.
2. `cudaMalloc` space for `arr2` and `cudaMemcpy` it.
3. Create a new array of pointers *on the GPU*, pointing to the new GPU addresses.

*This process is called a "Deep Copy" and it is incredibly slow and complex.*

---

### 3. A Sneaky Bug in your `matrix()` loop
Take a very close look at this line of your code:
```cpp
printf("%d ", *matrix[i]++);
```
This is a very clever C++ pointer trick, but it is **destructive**. 

Because of C++ operator precedence, `matrix[i]++` happens first. This permanently modifies the pointer stored inside your `matrix` array, pushing it forward by 1 memory slot. Then, the `*` dereferences it to print the number.

It works perfectly the first time you run it. But if you try to print the matrix a **second time** inside the same function:

```cpp
matrix(); // If you looped through it again inside this function...
```

It would print garbage values or crash your program! Why? Because your pointers `ptr1` and `ptr2` were moved to the end of the arrays during the first loop, and now point to empty space in your computer's RAM.

In `matrix_2d_array()`, you used `arr[i][j]`. This simply calculates an offset (the Golden Formula: `row * 4 + col`) and reads the value, leaving the original array completely untouched and safe to read again.

### Summary

* When a 2D array is created with `arr[][4]`, the compiler allocates a single block of contiguous memory for all the elements. Basically flat. This is perfect for hardware, caches, and GPUs. And to access the element the compiler uses the Golden Formula: `(row * COLS) + col` for the indexing.

* `matrix_2d_array()` = **Contiguous memory**. Perfect for hardware, caches, and GPUs. (Though in CUDA, we do it flat: `std::vector<int> arr = {1,2,3,4,5...}`).

* `matrix()` = **Array of pointers**. Fragmented memory. Good if you want a "jagged array" (where row 1 has 4 items, but row 2 has 99 items), but terrible for high-performance GPU computing.

## **Custom Types**

In C++, we use `size_t` to represent sizes and counts. 

- It is an `unsigned integer of long` type that is guaranteed to be large enough to hold the size of any object in memory.

- It has size of `64` bits on a 64-bit system, and 32 bits on a 32-bit system.

**Understand with example:**

If we've to create an array to hold millions or billions of integers, we can use `size_t` to represent the size of the array, rather than using a regular `int` which may not be able to hold such large values. For example:

```cpp
size_t N = 1 << 30; // ~1 Billion, instead of using int which can only hold up to 2^31 - 1
size_t N = 1 << 31; // ~2 Billion (CRASH!)
```

**Why does `1 << 31` fail?**

In C++, the number `1` is treated as a standard 32-bit signed `int` by default. 

Even though you are saving the result into a massive 64-bit `size_t`, the compiler performs the bitshift *first*. 

When it shifts the 32-bit `1` by 31 places, it hits the signed integer limit, overflows into a negative number, and *then* assigns that broken negative number to your `size_t`. 

### **Struct**

- Default size of `Struct` is the sum of the sizes of its members, plus any padding added for alignment. The size can vary based on the types of members and the compiler's alignment requirements.

Another way of creating custom types in C++ is using `Structs`. A struct is a user-defined data type that can hold multiple variables of different types. 

For example:

```cpp
struct Point {
    int x; // x-coordinate
    int y; // y-coordinate
};

Point p1; // create an instance of Point
p1.x = 10; // assign value to x
p1.y = 20; // assign value to y
```

In this example, we defined a struct called `Point` that has two integer members, `x` and `y`. We then created an instance of `Point` called `p1` and assigned values to its members.

> Difference between `Structs` and `Classes` in C++:

- In C++, the only difference between a `struct` and a `class` is the default access level. In a `struct`, members are public by default, while in a `class`, members are private by default.

## **Type Casting**

Type casting is the process of converting a variable from one data type to another. 

In standard C, casting was a "wild west" where you could force almost anything to be anything. C++ introduced **four specific, named casts** to make programming safer, more readable, and strict.

Since we're diving into High-Performance Computing (CUDA), understanding exactly how bytes are reinterpreted in memory is crucial.

> Notes:

- If we cast `int` to `char`, the compiler will take the 4 bytes of the integer and reinterpret them as a single byte (the first byte). This will give you the ASCII character corresponding to that byte. For example, if the integer is `65`, which is `0x00000041` in hexadecimal, casting it to `char` will give you `'A'`, because `0x41` is the ASCII code for 'A'.

---

### 1. Implicit Casting (Automatic)

You don't write any code for this. The compiler does it automatically when it knows it is safe, or when it tries to help you out.

```cpp
int a = 10;
double b = a;  // Implicitly casts int to double (10 becomes 10.0)

double x = 9.99;
int y = x;     // Implicitly casts double to int. (y becomes 9. Decimals are truncated!)
```

* **Warning:** Going from a large type (`double`) to a small type (`int`) is called a **Narrowing Conversion**. The compiler will usually throw a warning because you are losing data (the `.99` is destroyed).

---

### 2. C-Style Casting (The Old Way - Avoid in C++)
If you have written C, you know this one. You put the target type in parentheses. 

```cpp
double pi = 3.14159;
int int_pi = (int)pi; 
```
**Why we avoid this in C++:** 
A C-style cast is basically a sledgehammer. When you use it, the compiler tries multiple different ways to convert the data. It might do a safe mathematical conversion, or it might ruthlessly reassign the memory bits, causing catastrophic bugs. It doesn't communicate your *intent* to the compiler or to other programmers.

Because of this, C++ introduced the **4 Named Casts**.

---

### 3. The 4 Named C++ Casts

The syntax for all C++ casts looks like this: `cast_name<NewType>(variable);`

#### A. `static_cast` (The Workhorse)
This is the cast you will use **90% of the time**. It does safe, compile-time conversions. It asks the compiler, *"Hey, is there a logical, mathematical way to convert this to that?"*

```cpp
double d = 3.14;
// Safely converts double to int. Explicitly tells the compiler "I know I am losing the decimals."
int i = static_cast<int>(d); 

// In pure C++, standard malloc returns a 'void*' (a raw memory address with no type).
// We use static_cast to safely tell the compiler it is an array of integers.
int* arr = static_cast<int*>(malloc(10 * sizeof(int))); 
```
* **Use case:** Standard math conversions (`float` to `int`), casting pointers up and down standard inheritance trees, or converting `void*` to a typed pointer.

#### B. `reinterpret_cast` (The Dangerous One)
This tells the compiler: *"Do not change the bits in memory. Just put on a different pair of glasses and pretend those exact same bits are a completely different data type."*

This is heavily used in Systems Programming, game engines, and CUDA when you need to send raw bytes over a network or PCIe bus.

```cpp
int x = 65; 
// Pretend the memory address of this integer is actually a pointer to a character.
char* p = reinterpret_cast<char*>(&x);

// Because 65 is the ASCII code for 'A', this will print 'A'!
std::cout << *p << "\n"; 
```
* **Use case:** Low-level bit manipulation, hashing, converting a pointer into a `size_t` (an integer) so you can do math on the memory address, or hardware hacking. 
* *Note:* It does absolutely zero safety checks. If you `reinterpret_cast` an object wrong, your program will segfault instantly.

#### C. `const_cast` (The Hacker)
Its only job is to add or remove the `const` (read-only) lock from a variable. 

```cpp
void changeValue(const int* ptr) {
    // ptr is read-only. We cannot do *ptr = 50; 
    
    // We strip away the read-only lock
    int* unlocked_ptr = const_cast<int*>(ptr); 
    *unlocked_ptr = 50; // Now we can change it!
}

int main() {
    int x = 10;
    changeValue(&x);
    // x is now 50
}
```
* **Use case:** You should almost never use this. It is a "code smell." It is mostly used when you have a `const` variable, but you need to pass it into a very old, legacy 1990s C library that forgot to mark its parameters as `const`. 

#### D. `dynamic_cast` (The OOP Safety Net)
This is the only cast that happens at **Run-Time** (while the game/program is actively running), not Compile-Time. It is exclusively used for Object-Oriented Programming (Classes, Inheritance, and Polymorphism).

It safely checks if a "Base" class pointer is actually pointing to a specific "Derived" class. If the cast is invalid, it doesn't crash; it safely returns a `nullptr` (0).

```cpp
class Animal { virtual void speak() {} }; // Must have a virtual function!
class Dog : public Animal { public: void bark() { std::cout << "Woof!\n"; } };
class Cat : public Animal {};

int main() {
    Animal* my_animal = new Dog(); // It's an animal, but specifically a Dog

    // We want to make it bark. We use dynamic_cast to check if it's REALLY a dog.
    Dog* my_dog = dynamic_cast<Dog*>(my_animal);
    
    if (my_dog != nullptr) {
        my_dog->bark(); // Success!
    } else {
        std::cout << "This is not a dog!\n";
    }
    
    return 0;
}
```
* **Use case:** Safely navigating complex class hierarchies in UI frameworks or Game Engines.
* **CUDA Note:** `dynamic_cast` requires **RTTI** (Run-Time Type Information). Because RTTI takes up extra memory and slows down the system, **it is generally completely disabled on GPUs**. You will essentially never use `dynamic_cast` inside a CUDA `__global__` kernel.

## **Macros and Global Constants**

In C++ and CUDA, when you want to define a value or a piece of data that can be seen by your *entire* program, you have three main tools: **Macros**, **Global Variables**, and **Global Constants**. 

Although they might seem to do the same thing, they operate at completely different stages of the computer's pipeline (Pre-compilation vs. Run-time). 

Because you are learning CUDA—where you have thousands of threads running simultaneously—understanding the difference between these is absolutely critical. Here is the in-depth breakdown.

---

### 1. Macros (`#define`)
A macro is not actually part of the C++ language itself. It belongs to the **C Preprocessor**. 

Before your compiler (`g++` or `nvcc`) even looks at your code, the preprocessor runs through your file and does a literal **"Find and Replace"** on the text. 

**Syntax & Example:**
```cpp
#include <iostream>

#define PI 3.14159
#define NUM_THREADS 256
#define MULTIPLY(a, b) (a * b) // Function-like macro

int main() {
    double circle = PI * 2; 
    int total = MULTIPLY(10, 5);
    return 0;
}
```

**What happens under the hood:**
The compiler *never* sees the word `PI` or `MULTIPLY`. Before compiling, the preprocessor blindly rewrites your code to look like this:
```cpp
int main() {
    double circle = 3.14159 * 2; 
    int total = (10 * 5);
    return 0;
}
```

**Pros:**

* **Zero Memory Overhead:** Because it is just text replacement, macros don't take up any RAM.

* **CUDA Friendly:** Because it's just text, a `#define` works perfectly in both CPU and GPU code without needing `cudaMemcpy`. 

**Cons (Why modern C++ hates them):**

* **No Type Safety:** The compiler doesn't know if `PI` is a float, double, or string. It just pastes it.

* **Debugging Nightmare:** If a macro causes an error, the debugger points to the expanded text, not the word `PI`, which makes finding bugs very hard.

---

### 2. Global Variables
A global variable is a standard variable declared **outside** of any function. 

**Syntax & Example:**
```cpp
#include <iostream>

int global_counter = 0; // Lives in RAM permanently

void increment() {
    global_counter++; // Any function can touch this
}

int main() {
    increment();
    increment();
    std::cout << global_counter; // Prints 2
    return 0;
}
```

**What happens under the hood:**

When you compile the program, the compiler sets aside a permanent 4-byte chunk of your CPU RAM (in a section called the "Data Segment"). That memory address stays alive for the entire life of the program.

**Why Global Variables are TERRIBLE for CUDA:**
Global variables are generally considered bad practice in standard C++ because any function can secretly change the value, making bugs hard to track. 

But in CUDA, they are a disaster. Imagine you have a global variable `int counter = 0;`. You launch a CUDA kernel with **100,000 threads**. All 100,000 threads try to do `counter++` at the exact same nanosecond. 
* This causes a **Race Condition**. They overwrite each other, and instead of the counter reaching 100,000, it might randomly output 4,392. 
* In parallel programming, mutable (changeable) global state is your worst enemy.

---

### 3. Global Constants (`const` and `constexpr`)
This is the **modern C++ best practice**. It is a global variable that is locked as **Read-Only**. 

**Syntax & Example:**
```cpp
#include <iostream>

// const means "I promise not to change this at run-time"
const int MAX_USERS = 100;

// constexpr (C++11) means "Compute this at compile-time"
constexpr double PI = 3.1415926535;

int main() {
    double math = PI * 2;
    // MAX_USERS = 200; // COMPILER ERROR! Cannot modify read-only memory.
    return 0;
}
```

**What happens under the hood:**
The compiler sees a strict type (`double`) so it can enforce safety. Because it is marked `const`, the compiler might store it in a special Read-Only section of your computer's memory, or it might just optimize it away completely (acting just as fast as a macro, but with 100% type safety).

### How this applies to your CUDA Code

When you watch CoffeeBeforeArch, you will see a mix of these depending on what he is doing. Here is how they are used in real HPC (High-Performance Computing):

#### 1. When do we use Macros in CUDA?
We use them for **Error Checking**. In CUDA, almost every developer writes a macro called `CUDA_CHECK` to wrap their API calls. 
```cpp
#define CUDA_CHECK(call) \
    if((call) != cudaSuccess) { \
        printf("CUDA Error at line %d\n", __LINE__); \
    }

// Usage:
CUDA_CHECK(cudaMalloc(&d_A, bytes));
```
Because macros can read the `__LINE__` of your text file, they are incredible for instantly finding which `cudaMalloc` failed.

#### 2. When do we use Global Constants in CUDA?
When you have read-only data that every single GPU thread needs to see (like the gravity constant, or the dimensions of a matrix), you use a special CUDA feature called **Constant Memory**.

```cpp
// __constant__ tells the GPU to put this in ultra-fast, read-only cache memory
__constant__ float GRAVITY = 9.81f; 

__global__ void physicsKernel(float* velocities) {
    // All 10,000 threads can read GRAVITY simultaneously at lightning speed!
    velocities[threadIdx.x] += GRAVITY; 
}
```

## **Compilers**

A compiler is a special program that takes your human-readable source code (like `main.cpp`) and translates it into machine code (binary instructions) that your computer's CPU or GPU can execute directly. 

When you write C++ code, you typically use a compiler like `g++` for CPU code or `nvcc` for CUDA GPU code.

For `C` programs, you would use `gcc` instead of `g++`, but the concepts are similar. [Link](https://www.freecodecamp.org/news/what-is-a-compiler-in-c/)

For `C++` programs, you would use `g++` instead of `gcc`, but the concepts are similar. [Link](https://www.freecodecamp.org/news/c-compiler-explained-what-is-the-compiler-and-how-do-you-use-it/)

For `CUDA` programs, you would use `nvcc`, which is NVIDIA's CUDA Compiler. It can compile both CPU and GPU code, and it handles the complexities of generating code for the GPU architecture.
