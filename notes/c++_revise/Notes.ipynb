## **Pointer Notes**

> Special Notes

- When `*` asterisk is used in a `Declaration` it is used to declare a pointer variable. For example, `int* ptr;` declares a pointer to an integer.

- When `*` asterisk is used in an `Expression` it is used to dereference a pointer, which means accessing the value stored at the memory address that the pointer points to. For example, `int value = *ptr;` retrieves the value that `ptr` points to.

<hr>

Pointers are variables that store memory addresses. They allow us to directly access and manipulate memory, which can lead to more efficient code. In C/C++, we can declare a pointer using the `*` operator. For example:

```cpp
int* ptr; // declares a pointer to an integer
```
We can also initialize a pointer to point to a specific variable:

```cpp
int x = 10;
int* ptr = &x; // ptr now points to the memory address of x
```

To access the value stored at the memory address that a pointer points to, we can use the dereference operator `*`. For example:

```cpp
int value = *ptr; // value now holds the value of x, which is 10
```



## **Double Pointers**

Double pointers are pointers that point to other pointers. They are declared using two `*` operators. For example:

```cpp
#include <stdio.h>
using namespace std;

int main() {
    int x = 42;
    int* ptr = &x; // Stores the pointer to x 
    int** ptr2 = &ptr; // 
    int***ptr3 = &ptr2;

    printf("Value: %d\n", ***ptr3);
}
```

To understand how this code works and why we need multiple asterisks (`*`, `**`, `***`), it helps to visualize how computer memory works. 

In C and C++, every variable has a **value** and a **memory address** (where that value is stored). A pointer is simply a special variable that stores the memory address of another variable.

---

### Step 1: Visualizing the Memory

Let's look at what happens in memory step-by-step when your code runs. We will use hypothetical memory addresses (like `0x100`, `0x200`, etc.) to make it clear:

| Variable | Type | Value (What it stores) | Address of this variable | What it points to |
| :--- | :--- | :--- | :--- | :--- |
| **`x`** | `int` | `42` | `0x100` | *(An actual integer value)* |
| **`ptr`** | `int*` | `0x100` *(Address of x)* | `0x200` | Points to `x` |
| **`ptr2`** | `int**` | `0x200` *(Address of ptr)* | `0x300` | Points to `ptr` |
| **`ptr3`** | `int***` | `0x300` *(Address of ptr2)* | `0x400` | Points to `ptr2` |

---

### Step 2: Why can't we use a single asterisk (`*`) for everything?

In C and C++, the compiler is strictly typed. It needs to know exactly what kind of data is at the destination of a pointer. 

* **`int*`** means: *"This variable stores the address of an `int`."*
* **`int**`** means: *"This variable stores the address of an `int*` (a pointer to an integer)."*
* **`int***`** means: *"This variable stores the address of an `int**` (a pointer to a pointer to an integer)."*

If you tried to write:
```cpp
int* ptr2 = &ptr; // Error!
```
The compiler would reject this. Why? 
Because `ptr` is an `int*`. The address of `ptr` (`&ptr`) is of type `int**`. If you try to store it in a single-pointer `int* ptr2`, the compiler gets confused. It thinks `ptr2` is pointing directly to an integer, when in reality it is pointing to another pointer. 

Each extra asterisk tells the compiler, **"You need to take one more jump through memory to reach the actual integer."**

---

### Step 3: How Dereferencing Works (`***ptr3`)

The asterisk (`*`) has two different meanings depending on where it is used:
1. **In a declaration** (like `int* ptr`), it means "this is a pointer type."
2. **In an expression** (like `*ptr`), it is the dereference operator, which means "go to the address stored in this pointer and grab the value."

When you print `***ptr3`, you are dereferencing it three times. Think of it like peeling layers of an onion:

1. **`ptr3`** holds the address `0x300` (which is where `ptr2` lives).
2. **`*ptr3`** (First dereference): Go to `0x300`. The value stored there is `0x200` (which is the address of `ptr`).
3. **`**ptr3`** (Second dereference): Go to `0x200`. The value stored there is `0x100` (which is the address of `x`).
4. **`***ptr3`** (Third dereference): Go to `0x100`. The value stored there is `42`.

This is why the output of your program is `42`. If you only used `*ptr3`, it would print the memory address stored in `ptr2`. If you used `**ptr3`, it would print the memory address stored in `ptr`. To get the final integer, you must dereference it three times.




### **Pointer Arithmetic**

Pointer arithmetic allows us to perform operations on pointers, such as incrementing or decrementing them. When we increment a pointer, it moves to the next memory address based on the size of the data type it points to. For example:

```cpp
ptr++; // moves the pointer to the next integer memory address
```

### **Pointers and Arrays**

In C/C++, arrays and pointers are closely related. An array name can be treated as a pointer to the first element of the array. For example:

```cpp
int arr[5] = {1, 2, 3, 4, 5};
int* ptr = arr; // ptr points to the first element of the array
```

### **Pointers and Functions**

Pointers can also be used to pass arguments to functions by reference. This allows the function to modify the original variable. For example:

```cpp
void increment(int* num) {
    (*num)++; // increments the value at the memory address pointed to by num
}

int main() {
    int x = 5;
    increment(&x); // passes the address of x to the function
    // x is now 6
}
```

### **Pointers and Dynamic Memory Allocation**

> Normally when we compile `C++` program, all the memory is allocated at compile time, and the size of the memory is fixed. However, sometimes we may need to allocate memory dynamically at runtime, which is where pointers come in handy. This is called `Dynamic Memory Allocation`.

Pointers are essential for dynamic memory allocation, which allows us to allocate memory at runtime. In C/C++, we can use the `new` operator to allocate memory for a variable or an array. For example:

```cpp
int* ptr = new int; // allocates memory for a single integer
*ptr = 10; // assigns a value to the allocated memory
int* arr = new int[5]; // allocates memory for an array of 5 integers
```

> If we create a pointer using `new`, it is stored in the `Heap` memory, and even if it is declared inside a function, it will not be destroyed when the function ends. We need to manually free the allocated memory using the `delete` operator to avoid memory leaks:

```cpp
delete ptr; // frees the memory allocated for the single integer
delete[] arr; // frees the memory allocated for the array of integers
```

## **Void Pointers**

A void pointer, also known as a generic pointer, is a special type of pointer that can point to any data type. It is declared using the `void*` syntax. For example:

```cpp
void* ptr; // declares a void pointer
```

> Previously when we created a pointer, we had to specify the data type it pointed to. However, with void pointers, we can point to any data type without specifying it. This makes void pointers very flexible and useful in situations where we want to write functions that can work with different types of data.

To use a void pointer, we need to cast it to the appropriate data type before dereferencing it. For example:

```cpp
int x = 10;
void* ptr = &x; // ptr points to the memory address of x  

int value = *(int*)ptr; // casts ptr to an int pointer and dereferences it to get the value of x
```

Note: 

`malloc` returns a void pointer, so we need to cast it to the appropriate data type when using it in C++. In C, the cast is not necessary. For example:

```cpp
int* ptr = (int*)malloc(sizeof(int)); // allocates memory for a single integer
*ptr = 10; // assigns a value to the allocated memory
```

## **Null Pointers**

Null pointers are pointers that do not point to any valid memory address. They are often used to indicate that a pointer is not currently pointing to anything. In C/C++, we can assign a null pointer using the `nullptr` keyword (introduced in C++11) or the `NULL` macro (in C). For example:

```cpp
int* ptr = nullptr; // ptr is a null pointer
```
We cannot dereference a null pointer, as it does not point to a valid memory address. Attempting to do so will result in undefined behavior, which can lead to crashes or other unexpected results. Therefore, it is important to check if a pointer is null before dereferencing it:

```cpp
if (ptr != nullptr) {
    // safe to dereference ptr
} else {
    // ptr is null, handle accordingly
}
```

To derefence a null pointer, we can use the `*` operator, but it will lead to undefined behavior. For example:

```cpp
int value = *ptr; // this will cause undefined behavior since ptr is a null pointer
```

Therefore, to successfully dereference a null pointer, we need to type cast it to `void*` first, and then cast it back to the appropriate data type. For example:

```cpp
int value = *(int*)(void*)ptr; // this will still cause undefined behavior, but it demonstrates how to cast a null pointer to void* and back to int*
```
