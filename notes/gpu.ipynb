## **Brief Intro to GPU**

**CPU**

- General Purpose

- High Clock Speed 

- Few Cores

- High Cache 

- Low Latency 

- Low Throughput, less Operations per Cycle

**GPU**

- Specialized for Parallel Processing

- Low Clock Speed 

- Many Cores

- Low Cache

- High Latency

- High Throughput, more Operations per Cycle

## **CUDA** 

- CPU Allocates CPU Memory i.e. RAM 

- CPU copies data to GPU 

- CPU launches kernel on GPU (GPU executes the kernel)

- CPU copies results from GPU back to CPU to do something useful with it

### **Terms to Remember**

**Kernel**: A function that runs on the GPU. It is executed by many threads in parallel. Not the `Convolutional Neural Network` kernel, but the `CUDA` kernel.

**Threads**: The smallest unit of execution on the GPU. Each thread executes a kernel function. Threads are organized into blocks and grids.

**Blocks**: A group of threads that can cooperate with each other by sharing data through shared memory and synchronizing their execution. Each block can contain a certain number of threads (e.g., 256 or 512).

**Grids**: A collection of blocks. The grid size determines how many blocks are launched to execute a kernel. The total number of threads is the product of the number of blocks and the number of threads per block.

**GEMM (General Matrix Multiply)**: A common operation in linear algebra and deep learning, where two matrices are multiplied together. It is often used in neural network computations.

**SGEMM (Strided General Matrix Multiply)**: A variant of GEMM where the matrices are stored in a strided format, meaning that the elements of the matrices are not stored contiguously in memory. This can be useful for certain applications where the data is not naturally stored in a contiguous format.

> `CPU` is called the `host`, and `GPU` is called the `device`. The CPU (host) manages the GPU (device) and orchestrates the execution of kernels on the GPU.

### **NVidia GPU Architecture**

**Volta** : Volta is a GPU architecture developed by NVIDIA, released in 2017. It is designed for high-performance computing and deep learning applications. Volta introduced several new features, including Tensor Cores for accelerating deep learning workloads, and improved memory bandwidth. It is used in NVIDIA's Tesla V100 GPU.

**Ampere** : Ampere is the successor to Volta, released in 2020. It builds upon the features of Volta and introduces new ones, such as improved Tensor Cores, support for hardware-accelerated ray tracing, and increased performance for gaming and professional graphics applications. Ampere is used in NVIDIA's GeForce RTX 30 series GPUs.

**Ada Lovelace** : Ada Lovelace is the latest GPU architecture from NVIDIA, announced in 2022. It is designed for high-performance gaming and professional graphics applications. Ada Lovelace introduces new features such as improved ray tracing capabilities, enhanced AI performance, and support for next-generation gaming technologies. It is used in NVIDIA's GeForce RTX 40 series GPUs.

**Hopper** : Hopper is a GPU architecture developed by NVIDIA, announced in 2022. It is designed for high-performance computing and artificial intelligence workloads. Hopper introduces new features such as improved Tensor Cores, support for hardware-accelerated ray tracing, and increased performance for data center applications. It is used in NVIDIA's A100 GPU.
